## Infix to Postfix


```python
class Solution:
    def __init__(self):
        self.st = []

        self.priority = {
        "^": 3,
        "/":2,
        "*":2,
        "+":1,
        "-":1,
    }

    def push(self,x):
        self.st.append(x)

    def pop(self):
        self.st.pop()

    def top(self):
        return self.st[-1]
    
    def st_isempty(self):
        return len(self.st)==0

    def infixToPostfix(self, s: str) -> str:
        ans = ""

        for i in range(len(s)):

            if(('A' <= s[i] <= 'Z') or ('a' <= s[i] <= 'z') or ('0' <= s[i] <= '9')):
                ans = ans + s[i]
            
            elif(s[i]=="("):
                self.push(s[i])

            elif(s[i]==")"):
                while (not(self.st_isempty()) and (self.top() != "(")):
                    ans = ans+self.top()
                    self.pop()
                self.st.pop()

            else:
                while (not (self.st_isempty()) and self.top() != "(" and (self.priority[s[i]]<=self.priority[self.top()])):
                    ans = ans+self.top()
                    self.pop()
                self.push(s[i])

        while (not self.st_isempty()):
            ans = ans+ self.top()
            self.pop()


        return ans 
```

```python
from typing import List
from collections import deque 
class Solution:
    def isCycle(self, V: int, adj: List[List[int]]) -> bool:
        self.visited = set()
        q = deque()
        adj_list = {i:adj[i] for i in range (V)}

        q.append([0,-1])
        self.visited.add(0)

        while(len(q)>0):
            v,par = q.popleft()
            
            for neigh in adj_list[v]:

                if neigh not in self.visited:
                    q.append([neigh,v])
                    self.visited.add(neigh)

                elif neigh != par:
                    return True 

        return False 
```

```python
class Solution:
    def solve(self, board: List[List[str]]) -> None:
        visited = set()
        n = len(board)
        m = len(board[0])

        def dfs(i,j, board, visited):

            visited.add((i,j))

            if (i-1>=0 and (i-1,j) not in visited and board[i-1][j]=='O'):
                dfs(i-1,j,board,visited)
            if (j-1>=0 and (i,j-1) not in visited and board[i][j-1]=='O'):
                dfs(i,j-1,board,visited)
            if (i+1<n and (i+1,j) not in visited and board[i+1][j]=='O'):
                dfs(i+1,j,board,visited)
            if (j+1<m and (i,j+1) not in visited and board[i][j+1]=='O'):
                dfs(i,j+1,board,visited)

            return 

        for i in range (n):
            if ((i,0) not in visited and board[i][0]=='O'):
                dfs(i, 0, board, visited)

            if ((i,m-1) not in visited and board[i][m-1]=='O'):
                dfs(i,m-1,board, visited)

        for j in range (m):
            if ((0,j) not in visited and board[0][j]=='O'):
                dfs(0,j,board,visited)
            
            if ((n-1,j) not in visited and board[n-1][j]=='O'):
                dfs(n-1,j,board,visited)

        for i in range(n):
            for j in range(m):
                if board[i][j]=="O" and (i,j) not in visited:
                    board[i][j]='X'
```

## Bipartite Graph

```python
class Solution:
    def isBipartite(self, graph: List[List[int]]) -> bool:
        q = deque()
        n = len(graph)
        colour = [-1 for i in range (n)]

        def bfs(src,colour,q,graph):

            colour[src] = 1
            q.append(src)

            while(len(q)>0):
                i = q.popleft()
                for neigh in graph[i]:
                    if (colour[i]==1 and colour[neigh]==-1):
                        colour[neigh]=0
                        q.append(neigh)
                    elif (colour[i]==0 and colour[neigh]==-1):
                        colour[neigh]=1
                        q.append(neigh)
                    elif (colour[neigh]==0 or colour[neigh]==1):
                        if colour[neigh] == colour[i]:
                            return False

            return True 
            
        for i in range (n):
            if colour[i]==-1:
                if not bfs(i,colour,q,graph):
                    return False

        return True
```

## TOPO SORT

```python

class Solution:
    def topoSort(self, V, adj):
        stack = []
        n = V
        vis = set()
        out = []

        def dfs(i,stack,vis,adj):
            vis.add(i)
            for neigh in adj[i]:
                if neigh not in vis:
                    dfs(neigh,stack,vis,adj)
            stack.append(i)

        for i in range (n):
             if i not in vis:
                dfs(i,stack,vis,adj)

        while stack:
            out.append(stack.pop())
        
        return out
        
```

## Course Schedule 2

```python
class Solution:
    def findOrder(self, numCourses: int, prerequisites: List[List[int]]) -> List[int]:
        stack = []
        n = numCourses
        vis = set()
        rec_pth = set()
        out = []
        adj = {i:[] for i in range (n)}

        for u,v in prerequisites:
            adj[u].append(v)

        def dfs(i,stack,vis,adj):
            vis.add(i)
            rec_pth.add(i)
            for neigh in adj[i]:

                if neigh not in vis:
                    if not dfs(neigh,stack,vis,adj):
                        return False

                elif neigh in rec_pth:
                    return False

            rec_pth.remove(i)
            stack.append(i)

            return True

        for i in range (n):
             if i not in vis:
                if not dfs(i,stack,vis,adj):
                    return []

        while stack:
            out.append(stack.pop())
        
        return out[::-1]
        
```

## EVENTUAL SAFE STATE

```python
class Solution:
    def eventualSafeNodes(self, graph: List[List[int]]) -> List[int]:
        visited = set()
        rec_path = set()
        n = len(graph)
        check = {}
        output = []

        def dfs(i,visited,graph,check,rec_path):
            visited.add(i)
            rec_path.add(i)
            check[i] = 0

            for neigh in graph[i]:

                if neigh not in visited:
                    if dfs(neigh,visited,graph,check,rec_path)==True:
                        return True

                elif neigh in rec_path:
                    return True

            rec_path.remove(i)
            check[i] = 1
            return False

        for i in range (n):
            if i not in visited:
                dfs(i,visited,graph,check,rec_path)

        for i in range (n):
            if check[i]==1:
                output.append(i)

        return output


        
```

# DIJKSTRA ALGO USING HEAP/PQ

```python
import heapq
import math

class Solution:
    def dijkstra(self, V, edges, S):
        x = math.inf
        pq = []
        adj_list = {i:[] for i in range (V)}

        for u, v, w in edges:
            adj_list[u].append([v,w])
            adj_list[v].append([u,w])

        src = S
        dis = [x for i in range (V)]
        dis[src] = 0
        heapq.heappush(pq,(0,src))

        while(pq):
            weight,node = heapq.heappop(pq)
            for neigh in adj_list[node]:
                if dis[neigh[0]]>neigh[1]+weight:
                    dis[neigh[0]]=neigh[1]+weight
                    heapq.heappush(pq,(dis[neigh[0]],neigh[0]))
        
        return dis
```

```python
import math

class Solution:
    def shortestPathBinaryMatrix(self, grid: List[List[int]]) -> int:
        n = len(grid)
        m = len(grid[0])
        dis = [[math.inf for i in range (m)] for i in range (n)]
        q = deque()
        q.append([0,0])
        dis[0][0] = 1

        if (grid[0][0]!=0 or grid[n-1][m-1]!=0) :
            return -1

        while(len(q) > 0):

            i,j = q.popleft()
            d = dis[i][j]

            if (i==n-1 and j==m-1):
                return dis[i][j]
                
            else:
                if (i+1<n and grid[i+1][j]==0 and dis[i+1][j]>d+1):
                    dis[i+1][j]=d+1
                    q.append([i+1,j])
                if (j+1<m and grid[i][j+1]==0 and dis[i][j+1]>d+1):
                    dis[i][j+1]=d+1
                    q.append([i,j+1])
                if (i-1>=0 and grid[i-1][j]==0 and dis[i-1][j]>d+1):
                    dis[i-1][j]=d+1
                    q.append([i-1,j])
                if (j-1>=0 and grid[i][j-1]==0 and dis[i][j-1]>d+1):
                    dis[i][j-1]=d+1
                    q.append([i,j-1])
                if (i+1<n and j+1<m and grid[i+1][j+1]==0 and dis[i+1][j+1]>d+1):
                    dis[i+1][j+1]=d+1
                    q.append([i+1,j+1])
                if (i+1<n and j-1>=0 and grid[i+1][j-1]==0 and dis[i+1][j-1]>d+1):
                    dis[i+1][j-1]=d+1
                    q.append([i+1,j-1])
                if (i-1>=0 and j-1>=0 and grid[i-1][j-1]==0 and dis[i-1][j-1]>d+1):
                    dis[i-1][j-1]=d+1
                    q.append([i-1,j-1])
                if (i-1>=0 and j+1<m and grid[i-1][j+1]==0 and dis[i-1][j+1]>d+1):
                    dis[i-1][j+1]=d+1
                    q.append([i-1,j+1])
        
        return -1




        
```

# Number of Ways to arrive at distance (THERE ARE 2 to 3 more question not uploaded here pls check in sheet)

```python
import heapq
import math

class Solution:
    def countPaths(self, n: int, roads: List[List[int]]) -> int:
        adj_list = {i:[] for i in range (n)}
        dis = [math.inf for i in range (n)]
        ways = [0 for i in range (n)]

        for u,v,w in roads:
            adj_list[u].append([v,w])
            adj_list[v].append([u,w])

        pq = []
        heapq.heapify(pq)
        src = 0
        heapq.heappush(pq,[src,0])
        ways[src] = 1
        mod = 1e9 + 7

        while (len(pq)>0):
            d,src = heapq.heappop(pq)
            for neigh in adj_list[src]:
                if dis[neigh[0]]>d+neigh[1]:
                    dis[neigh[0]] = d+neigh[1]
                    heapq.heappush(pq,[dis[neigh[0]],neigh[0]])
                    ways[neigh[0]] = ways[src]
                
                elif dis[neigh[0]]==d+neigh[1]:
                    ways[neigh[0]] = int((ways[src]+ways[neigh[0]])%mod)

        return int ((ways[n-1])%mod)     


```

# MIN MULTIPLICATION TO REACH END 

```python
import math

class Solution:
    def minimumMultiplications(self, arr, start, end):
        q = deque()
        mod = int(1e5)
        time = [math.inf for i in range (mod)]
        time[start] = 0
        q.append([0,start])

        while(len(q)>0):
            t,node = q.popleft()

            if node==end:
                return time[node]

            for i in range (len(arr)):
                if time[(node*arr[i])%mod]>t+1:
                    q.append([t+1,(node*arr[i])%mod])
                    time[(node*arr[i])%mod]=t+1
                
        return -1
```

# BELLMAN FORD ALGO (SHORTEST DISTANCE OF ALL THE VERTICES FROM SRC)

```python
import math
class Solution:
    def bellman_ford(self, V, edges, S):

        dis = [1e9 for i in range (V)]

        dis[S] = 0

        for i in range (V-1):
            for u,v,w in edges:
                if dis[u]!=1e9 and dis[v]>dis[u]+w:
                    dis[v]=dis[u]+w

        for i in range (V-1):
            for u,v,w in edges:
                if dis[u]!=1e9 and dis[v]>dis[u]+w:
                    return -1

        return dis
```
