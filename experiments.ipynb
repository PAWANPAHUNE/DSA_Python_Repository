# DoublyLinkedList Constructor

```python
from Constructor.DoublyLinkedList import DoublyLinkedList, Node
dll = DoublyLinkedList(10)
```

```python
dll.print_list()
```

```python
dll.append(20)
dll.append(30)

dll.print_list()
```

```python
removed = dll.pop()
print("Removed:", removed.value)
dll.print_list()
```

```python
dll.prepend(5)
dll.print_list()
```

```python
dll.pop_first()
dll.print_list()
```

```python
print(dll.get(0).value)   # head
print(dll.get(1).value)   # tail
print(dll.get(2))         # out of bounds
```

```python
dll.set(1, 99)
dll.print_list()
```

```python
dll.insert(0, 1)
dll.print_list()
```
