# rBST Test Cases

```python
from Constructor.recursive_BST import rBST, Node

# Create BST
bst = rBST()

# Test recursive insert
values = [47, 21, 76, 18, 27, 52, 82]
for v in values:
    bst.r_insert(v)

print("Insert Test Completed")

# Test contain
print("Contain 27:", bst.r_contain(27))   
print("Contain 100:", bst.r_contain(100)) 

# Test minimum value in subtree
print("Min value from root:", bst.min_subtree(bst.root))  

# Test delete leaf node
bst.delete_node(18)
print("Contain 18 after delete:", bst.r_contain(18))  

# Test delete node with one child
bst.delete_node(27)
print("Contain 27 after delete:", bst.r_contain(27))  

# Test delete node with two children
bst.delete_node(47)
print("Contain 47 after delete:", bst.r_contain(47))  

print("All tests executed")
```
