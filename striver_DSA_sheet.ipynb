## Infix to Postfix


```python
class Solution:
    def __init__(self):
        self.st = []

        self.priority = {
        "^": 3,
        "/":2,
        "*":2,
        "+":1,
        "-":1,
    }

    def push(self,x):
        self.st.append(x)

    def pop(self):
        self.st.pop()

    def top(self):
        return self.st[-1]
    
    def st_isempty(self):
        return len(self.st)==0

    def infixToPostfix(self, s: str) -> str:
        ans = ""

        for i in range(len(s)):

            if(('A' <= s[i] <= 'Z') or ('a' <= s[i] <= 'z') or ('0' <= s[i] <= '9')):
                ans = ans + s[i]
            
            elif(s[i]=="("):
                self.push(s[i])

            elif(s[i]==")"):
                while (not(self.st_isempty()) and (self.top() != "(")):
                    ans = ans+self.top()
                    self.pop()
                self.st.pop()

            else:
                while (not (self.st_isempty()) and self.top() != "(" and (self.priority[s[i]]<=self.priority[self.top()])):
                    ans = ans+self.top()
                    self.pop()
                self.push(s[i])

        while (not self.st_isempty()):
            ans = ans+ self.top()
            self.pop()


        return ans 
```

## Connected Components in Graph 

```python
class Solution:
    def findNumberOfComponent(self, V, edges):
        visited = set()
        connected = 0

        def dfs(node):
            visited.add(node)
            for neighbour in graph[node]:
                if neighbour not in visited:
                    dfs(neighbour)
            return 

        graph = {node:[] for node in range(V)}

        for u,v in edges:
            graph[u].append(v)
            graph[v].append(u)
        
        for node in graph:
            if node not in visited:
                connected +=1
                dfs(node)
        return connected
```

## Cycle Detection in Undirected Graph (BFS)

```python
from typing import List
from collections import deque 
class Solution:
    def isCycle(self, V: int, adj: List[List[int]]) -> bool:
        self.visited = set()
        q = deque()
        adj_list = {i:adj[i] for i in range (V)}

        q.append([0,-1])
        self.visited.add(0)

        while(len(q)>0):
            v,par = q.popleft()
            
            for neigh in adj_list[v]:

                if neigh not in self.visited:
                    q.append([neigh,v])
                    self.visited.add(neigh)

                elif neigh != par:
                    return True 

        return False 
```

## Cycle Detection in Undirected Graph (DFS)

```python
from typing import List
from collections import deque 
class Solution:
    
    def isCycle(self, V: int, adj: List[List[int]]) -> bool:

        def dfs(vis, src, par, adj_list):
            vis.add(src)
            for neigh in adj_list[src]:
                if neigh not in vis:
                    if dfs(vis, neigh, src, adj_list):
                        return True
                elif neigh!= par:
                    return True

        self.visited = set()
        adj_list = {i:adj[i] for i in range (V)}

        for i in range (V):
            if i not in self.visited:
                if dfs(self.visited,i,-1,adj_list):
                    return True 

        return False
```

## Cycle detection in Directed Graph Using DFS 

```python
class Solution:
    def isCyclic(self, N, adj):
        adj_list = {i:adj[i] for i in range (N)}
        self.vis = set()
        self.rec_pth = set()

        def dfs(src,adj_list,vis,rec_pth):
            vis.add(src)
            rec_pth.add(src)

            for neigh in adj_list[src]:
                if (neigh not in vis):

                    if (dfs(neigh, adj_list, vis, rec_pth)):
                        return True

                elif (neigh in rec_pth):
                    return True

            rec_pth.remove(src)
        
        for i in range (N):
            if i not in self.vis:
                if (dfs(i, adj_list, self.vis, self.rec_pth)):
                    return True 

        return False

        
                    
```

## 01 Matrix

```python
class Solution:
    def updateMatrix(self, mat: List[List[int]]) -> List[List[int]]:
        q = deque()
        visited = set()
        n = len(mat)
        m = len(mat[0])
        dis = [[0 for i in range (m)] for i in range (n)]

        for i in range (n):
            for j in range (m):
                if mat[i][j]==0:
                    q.append([i,j,0])
                    visited.add((i,j))

        while(len(q)>0):
            i,j,d = q.popleft()
            dis[i][j] = d

            if (i-1>=0 and (i-1,j) not in visited and mat[i-1][j]==1):
                q.append([i-1,j,d+1])
                visited.add((i-1,j))

            if (i+1<n and (i+1,j) not in visited and mat[i+1][j]==1):
                q.append([i+1,j,d+1])
                visited.add((i+1,j))

            if (j-1>=0 and (i,j-1) not in visited and mat[i][j-1]==1):
                q.append([i,j-1,d+1])
                visited.add((i,j-1))

            if (j+1<m and (i,j+1) not in visited and mat[i][j+1]==1):
                q.append([i,j+1,d+1])
                visited.add((i,j+1))
        
        return dis

```

## sorrounded region


```python
class Solution:
    def solve(self, board: List[List[str]]) -> None:
        visited = set()
        n = len(board)
        m = len(board[0])

        def dfs(i,j, board, visited):

            visited.add((i,j))

            if (i-1>=0 and (i-1,j) not in visited and board[i-1][j]=='O'):
                dfs(i-1,j,board,visited)
            if (j-1>=0 and (i,j-1) not in visited and board[i][j-1]=='O'):
                dfs(i,j-1,board,visited)
            if (i+1<n and (i+1,j) not in visited and board[i+1][j]=='O'):
                dfs(i+1,j,board,visited)
            if (j+1<m and (i,j+1) not in visited and board[i][j+1]=='O'):
                dfs(i,j+1,board,visited)

            return 

        for i in range (n):
            if ((i,0) not in visited and board[i][0]=='O'):
                dfs(i, 0, board, visited)

            if ((i,m-1) not in visited and board[i][m-1]=='O'):
                dfs(i,m-1,board, visited)

        for j in range (m):
            if ((0,j) not in visited and board[0][j]=='O'):
                dfs(0,j,board,visited)
            
            if ((n-1,j) not in visited and board[n-1][j]=='O'):
                dfs(n-1,j,board,visited)

        for i in range(n):
            for j in range(m):
                if board[i][j]=="O" and (i,j) not in visited:
                    board[i][j]='X'
```
